## 📊 Key Business Insights

- Food and beverages was the best-selling product line, generating total sales of 56,144.84.
- Giza was the best-performing branch with total sales of 110,568.71.
- Member customers generated higher total sales of 189,694.76.
- January recorded the highest monthly sales of 116,291.87.
- Cash was the highest-selling payment method with total sales of 112,206.57.
- Female customers generated higher total sales than male customers.
- Electronic accessories had the highest quantity sold.
- Food and beverages generated the highest gross income.
- Food and beverages received the highest average customer rating.
